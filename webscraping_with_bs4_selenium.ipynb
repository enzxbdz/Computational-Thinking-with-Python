# IMDB Scraper

```python
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
}

url = "https://www.imdb.com/chart/top/?ref_=hm_nv_menu"

response = requests.get(url, headers=headers)
print(response.status_code)
print(response.text)
```

```python
soup = BeautifulSoup(response.text, 'html.parser')
```

```python
print(soup.find('title').text)
```

```python
soup.find(html)
```

```python
data = soup.find(class_='ipc-metadata-list ipc-metadata-list--dividers-none sc-d24d5d37-0 hDHQeM alt_item_colors compact-list-view ipc-metadata-list--base')
```

```python
titles = []
ratings = []
votes = []

for title in data.find_all(class_='ipc-title__text'):
    titles.append(title.text)

print(titles)
```

```python
for rating in data.find_all(class_='ipc-rating-star--rating'):
    ratings.append(rating.text)

print(ratings)
```

```python
for vote in data.find_all(class_='ipc-rating-star--voteCount'):
    votes.append(vote.text.split(")")[0][2:])

print(votes)
```

```python
import pandas as pd 

data_dict = {
    "titles": titles,
    "votes": votes,
    "ratings": ratings
}

df = pd.DataFrame(data_dict)
df
```

```python
numbers = {
    "K": 1_000,
    "M": 1_000_000
}

def votes_to_float(vote, numbers=numbers):
    unit = vote[-1]
    if unit in numbers:
        result = float(vote[:-1])
        result = result * numbers[unit]
    else:
        result = float(vote)
    return result

## test
votes_to_float("10K")
```
