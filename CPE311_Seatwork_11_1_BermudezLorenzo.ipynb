# Part 1:

```python
# Suppose we had a linear regression model with one feature that learned the following bias and weights:
b=1
w1=3

# Further, suppose the following feature values for a given example:
x1=0

# Compute the linear regression prediction for this particular example using python in Google Colab.
# Make sure to plot the solution.Provide an analysis for the output.
```

```python
# Calculation
prediction_part1 = (w1 * x1) + b

print(f"Part 1 Prediction: {prediction_part1}")
```

```python
# Plotting the linear relationship
import matplotlib.pyplot as plt
import numpy as np

x_range = np.array([-2, -1, 0, 1, 2])
y_range = [ (w1 * x) + b for x in x_range ]

plt.plot(x_range, y_range, label="y = 3x + 1")
plt.scatter(x1, prediction_part1, color='green', label=f"Point (0, {prediction_part1})")
plt.xlabel("x1")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```

## Analysis:

##### In this part, I follow the linear equation y=3x+1. When x1​=0, the weight's effect is nullified (3×0=0), which makes b=1 the only contributor to the prediction. Geometrically, this point is the y-intercept, showing where the line starts on the graph when the input is zero. The slope of 3 means that for every step you move right on the x-axis, the prediction climbs up by 3.

# Part 2:

```python
# Define parameters
weights = [3, -2, 4]
bias = 1
```

```python
# Provided example X values
x_sets = [
    [3, -1, 2],
    [10, 0, -10]
]
```

```python
def predict(x, w, b):
    # (w1*x1 + w2*x2 + w3*x3) + b
    y = (w[0]*x[0]) + (w[1]*x[1]) + (w[2]*x[2]) + b
    return y
```

```python
# Calculating predictions
results = [predict(x, weights, bias) for x in x_sets]

print(f"Prediction 1: {results[0]}")
print(f"Prediction 2: {results[1]}")
```

```python
# Plotting the regression predictions
labels = ['Set 1', 'Set 2']
plt.figure(figsize=(8, 5))
plt.bar(labels, results, color=['green', 'blue'], alpha=0.7)
plt.axhline(0, color='black', lw=1)
plt.ylabel("Predicted y")
plt.title("Part 2: Manual Regression Predictions")
plt.grid(axis='y', linestyle='--', alpha=0.5)

for i, v in enumerate(results):
    plt.text(i, v + (0.5 if v > 0 else -1.5), str(v), ha='center', fontweight='bold')

plt.show()
```

## Analysis:

##### In this part, I follow the linear equation y=(w1​x1​+w2​x2​+w3​x3​)+b to handle multiple features at once. For Example 1, the calculation (3×3)+(−2×−1)+(4×2)+1 results in 20; even though w2​ is negative, the negative input for x2​ ends up adding to the total. In Example 2, the result is (3×10)+(−2×0)+(4×−10)+1, which equals -9. Here, the large negative product from w3​ and x3​ completely outweighs the high value of x1​, pulling the final prediction below zero. This shows that in multivariate regression, the final output is a tug-of-war between different features depending on their specific weights.
